```python
import pandas as pd

# Create sample sales data
data = {
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Keyboard'],
    'Sales': [50000, 30000, 20000, 5000, 3000],
    'Quantity': [5, 10, 8, 20, 15]
}

# Convert the data into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

```python
# Calculate total revenue
total_revenue = df['Sales'].sum()

# Calculate total quantity sold
total_quantity = df['Quantity'].sum()

print("Total Revenue: ₹", total_revenue)
print("Total Quantity Sold:", total_quantity)
```

```python
df.to_csv("sales_data.csv", index=False)
```

```python
import pandas as pd

df = pd.read_csv("sales_data.csv")

print(df)
```

```python
total_sales = df["Sales"].sum()

average_sales = df["Sales"].mean()

highest_product = df.loc[df["Sales"].idxmax(), "Product"]

print("Total Sales: ₹", total_sales)
print("Average Sales: ₹", average_sales)
print("Highest Revenue Product:", highest_product)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(df["Product"], df["Sales"])

plt.title("Sales Revenue by Product")

plt.xlabel("Product")

plt.ylabel("Revenue (INR)")

plt.show()
```

Laptops generated the highest revenue.

Business insight: Laptops contribute approximately 46.3% of total revenue.

This suggests that laptops are an important revenue-generating product in this sample dataset.
